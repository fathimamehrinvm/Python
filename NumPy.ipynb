#linspace()
creates numbers that are equally spaced b/w start and end value

np.linspace(start,stop,num)

start - starting value

stop- end

num- how many values u want

* note- end number is included


```python
np.linspace(1,10,5)
```

```python
a=np.linspace(0,10,5, endpoint=False)
print(a)
```

```python
a=np.linspace(0,10,5, endpoint=True)
print(a)
```

#statistics and axis

```python
arr=np.array([10,20,30,40,50])
print(arr.sum())
print(arr.mean())
print(arr.min())
print(arr.max())
```

# AXIS
axis=0 means column-wise operation

axis=1 means row-wise operation


```python
arr=np.array([
    [1,2,3],
    [4,5,6]
])
print(arr.sum(axis=0))
print(arr.sum(axis=1))
```
